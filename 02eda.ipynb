```python
import pandas as pd
df = pd.read_csv("data/ecommerce_sessions.csv")
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
```

```python
df.columns.tolist()
```

```python
sns.countplot(x='Converted', data=df)
plt.title("Converted Distribution")
plt.show()
```

```python
df['Converted'].value_counts()
```

```python
df['Converted'].value_counts(normalize=True)*100
```

```python
num_cols = df.select_dtypes(include=['int64','float64']).columns

df[num_cols].hist(figsize=(15,12), bins=20)

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(15,10))

sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm")

plt.title("Correlation Matrix")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.heatmap(df.isnull(),
            cbar=False,
            cmap="viridis")

plt.show()
```

```python
for col in num_cols:
    plt.figure(figsize=(6,3))

    sns.boxplot(x=df[col])

    plt.title(col)

    plt.show()
```

```python
num_cols = num_cols.drop("Converted")
```

```python
for col in num_cols:

    plt.figure(figsize=(6,4))

    sns.boxplot(x='Converted',
                y=col,
                data=df)

    plt.title(col)

    plt.show()
```

```python
cat_cols = df.select_dtypes(include='object').columns

cat_cols
```

```python
for col in cat_cols:

    plt.figure(figsize=(8,4))

    sns.countplot(x=col,
                  data=df)

    plt.xticks(rotation=45)

    plt.title(col)

    plt.show()
```

```python
for col in cat_cols:

    plt.figure(figsize=(8,4))

    sns.countplot(x=col,
                  hue='Converted',
                  data=df)

    plt.xticks(rotation=45)

    plt.title(col)

    plt.show()
```

```python
sample_df = df.sample(500, random_state=42)

sns.pairplot(sample_df,
             hue='Converted')

plt.show()
```
